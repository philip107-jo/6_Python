### 8. 탑승객을 나이대 기준으로 그룹화하여 새로운 열 `AgeGroup`을 추가한 후, 상위 5개 행을 확인하시오.

**나이대 구분 기준**

| 나이 범위 | `AgeGroup` 값 |
| --- | --- |
| 0세 이상 ~ 10세 미만 | `'아동'` |
| 10세 이상 ~ 20세 미만 | `'10대'` |
| 20세 이상 ~ 30세 미만 | `'20대'` |
| 30세 이상 ~ 40세 미만 | `'30대'` |
| 40세 이상 ~ 50세 미만 | `'40대'` |
| 50세 이상 ~ 60세 미만 | `'50대'` |
| 60세 이상 | `'60대 이상'` |
| 결측치(`NaN`) | `'미확인'` |

- `pd.cut()` 또는 조건식(if-else / `np.where`)을 자유롭게 사용해도 됩니다.

```python
df['AgeGroup']=pd.cut(df['Age'],bins=[0,10,20,30,40,50,60,np.inf],labels=['아동', '10대', '20대', '30대', '40대', '50대', '60대 이상'],right=False)
df['AgeGroup']=df['AgeGroup'].cat.add_categories('미확인').fillna('미확인')
df.head()
```

### 9. 성별(`Sex`)과 객실 등급(`Pclass`)을 기준으로 그룹화하여 각각의 평균 생존율을 계산하시오.

```python
df_survival = df.groupby(['Sex','Pclass']).agg(
            평균생존율=("Survived","mean")
        )
df_survival.head()
```

### 10. 나이대별 평균 생존율을 계산하시오.
